```python
display(df.describe(include='all').T)
```

```python
# Create a separate working copy
df_clean = df.copy()

# Remove duplicate records only
df_clean = df_clean.drop_duplicates()

# Reset index
df_clean = df_clean.reset_index(drop=True)

print("Original dataset:", df.shape)
print("Cleaned dataset:", df_clean.shape)

display(df_clean.head())
```

```python
df_clean.to_csv(
    '/kaggle/working/cleaned_water_quality.csv',
    index=False
)
```

```python
# Display information about the cleaned dataset

print("Dataset Shape:", df_clean.shape)

print("\nColumn Information:")
df_clean.info()

print("\nStatistical Summary:")
display(df_clean.describe().T)
```

```python
# Important water quality parameters

water_columns = [
    'Temp',
    'D.O. (mg/l)',
    'PH',
    'CONDUCTIVITY (µmhos/cm)',
    'B.O.D. (mg/l)',
    'NITRATENAN N+ NITRITENANN (mg/L)',
    'FECAL COLIFORM (MPN/100ml)',
    'TOTAL COLIFORM (MPN/100ml)Mean'
]

# Check that column names exist
available_columns = [
    col for col in water_columns
    if col in df_clean.columns
]

print("Available water parameters:")
print(available_columns)
```

```python
eda_df = df_clean.copy()

for col in available_columns:
    eda_df[col] = pd.to_numeric(
        eda_df[col],
        errors='coerce'
    )

print("Numeric conversion completed!")

display(eda_df[available_columns].describe().T)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in available_columns:

    plt.figure(figsize=(8, 4))

    sns.histplot(
        data=eda_df,
        x=col,
        kde=True,
        bins=30
    )

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()
```

```python
for col in available_columns:

    plt.figure(figsize=(8, 3))

    sns.boxplot(x=eda_df[col])

    plt.title(f'Outlier Detection: {col}')
    plt.xlabel(col)

    plt.tight_layout()
    plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = eda_df[available_columns].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Between Water Quality Parameters')

plt.tight_layout()
plt.show()
```

```python
if 'STATE' in eda_df.columns:

    state_data = (
        eda_df.groupby('STATE')['PH']
        .mean()
        .sort_values(ascending=False)
        .head(10)
    )

    plt.figure(figsize=(12, 5))

    sns.barplot(
        x=state_data.index,
        y=state_data.values
    )

    plt.title('Top 10 States by Average Water pH')
    plt.xlabel('State')
    plt.ylabel('Average pH')
    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()
```
