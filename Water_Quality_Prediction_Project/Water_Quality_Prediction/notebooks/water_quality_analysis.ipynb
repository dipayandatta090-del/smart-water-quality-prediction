# Water Quality Prediction — Data Analysis
Run this notebook from the project root after adding `data/water_potability.csv`.

```
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = Path('../data/water_potability.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

## Dataset overview

```
print('Shape:', df.shape)
display(df.info())
display(df.describe())
display(df.isnull().sum())
```

## Target distribution

```
target = 'Potability'
if target in df.columns:
    sns.countplot(data=df, x=target)
    plt.title('Potability Class Distribution')
    plt.show()
else:
    print(f'Change target variable: {target}. Available columns: {list(df.columns)}')
```

## Correlation heatmap

```
numeric = df.select_dtypes(include='number')
plt.figure(figsize=(10, 7))
sns.heatmap(numeric.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()
```

## Missing values

```
missing = df.isnull().sum().sort_values(ascending=False)
missing[missing > 0]
```
